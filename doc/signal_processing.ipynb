```python
import numpy as np
import pandas as pd

import emd
```

```python
data = pd.read_csv('../data/raw/dataset_2019_2025.csv')
data['datetime'] = pd.to_datetime(data['datetime'], utc=True)
data = data[data['datetime'].dt.year >= 2024]
data.to_csv('../data/raw/dataset_2024_2025.csv', index=False)
print(data.shape)
data.head()
```

```python
price = data['price'].to_numpy()
imf = emd.sift.sift(price)
emd.plotting.plot_imfs(imf)
```

```python
imf = emd.sift.complete_ensemble_sift(price, nensembles=16)
emd.plotting.plot_imfs(imf)
```
